```python
import matplotlib.pyplot as plt

# Data
labels = ['Grade 1', 'Grade 2', 'Grade 3', 'Grade 4']
counts = [16, 787, 788, 292]

# Create the bar chart
plt.figure(figsize=(8, 6))
bars = plt.bar(labels, counts, color=['#4C72B0', '#55A868', '#C44E52', '#8172B3'], width=0.6)

# Add titles and labels
plt.title('Tumor Grade Distribution', fontsize=16)
plt.xlabel('Tumor Grade', fontsize=12)
plt.ylabel('Number of Samples', fontsize=12)

# Add value labels on top of the bars
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2.0, yval + 10, int(yval), ha='center', va='bottom', fontsize=11)

plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
